```python
import boto3

s3 = boto3.client("s3")

bucket_name = "dsan6000-bm1262"

s3.create_bucket(Bucket=bucket_name)

print(f"Created bucket: s3://{bucket_name}")
```

```python
response = s3.list_buckets()

for bucket in response["Buckets"]:
    print(bucket["Name"])
```

```python
from pathlib import Path

data_dir = Path("data")

for file_path in sorted(data_dir.glob("*.parquet")):
    s3_key = f"wikipedia-hourly/{file_path.name}"

    s3.upload_file(
        str(file_path),
        bucket_name,
        s3_key
    )

    print(f"Uploaded {file_path.name}")
```

```python
response = s3.list_objects_v2(
    Bucket=bucket_name,
    Prefix="wikipedia-hourly/"
)

objects = [
    obj for obj in response.get("Contents", [])
    if obj["Key"].endswith(".parquet")
]

print(f"Number of Parquet files in S3: {len(objects)}")
```

```python
total_size = sum(obj["Size"] for obj in objects)

print(f"Total size: {total_size / 1e6:.2f} MB")
```
